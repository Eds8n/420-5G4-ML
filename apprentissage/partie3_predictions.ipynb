# Pokémon — Partie 3 : prédictions

Charger le modèle entraîné dans `partie3_entrainement.ipynb` et prédire le résultat d’un combat.

```python
from pathlib import Path
import csv
import joblib

assert Path("modele/modele_pokemon.mod").exists(), "Exécuter d'abord partie3_entrainement.ipynb"
assert Path("datas/pokedex.csv").exists(), "Exécuter depuis le dossier apprentissage/"
```

## 1. Trouver les caractéristiques d’un Pokémon

```python
def rechercheInformationsPokemon(numPokemon, Pokedex):
    infosPokemon = []
    for pokemon in Pokedex:
        if int(pokemon[0]) == numPokemon:
            # Numéro, nom, puis les sept caractéristiques du tutoriel
            infosPokemon = [pokemon[0], pokemon[1], pokemon[4], pokemon[5],
                            pokemon[6], pokemon[7], pokemon[8], pokemon[9], pokemon[10]]
            break
    return infosPokemon
```

## 2. Comparer deux Pokémon

```python
def prediction(numeroPokemon1, numeroPokemon2, Pokedex):
    pokemon1 = rechercheInformationsPokemon(numeroPokemon1, Pokedex)
    pokemon2 = rechercheInformationsPokemon(numeroPokemon2, Pokedex)
    if not pokemon1 or not pokemon2:
        raise ValueError("Un des numéros de Pokémon est introuvable dans le Pokédex.")

    modele_prediction = joblib.load("modele/modele_pokemon.mod")
    prediction_Pokemon_1 = modele_prediction.predict([[float(valeur) for valeur in pokemon1[2:]]])
    prediction_Pokemon_2 = modele_prediction.predict([[float(valeur) for valeur in pokemon2[2:]]])

    print(f"COMBAT OPPOSANT : ({numeroPokemon1}) {pokemon1[1]} à ({numeroPokemon2}) {pokemon2[1]}")
    print(f"{pokemon1[1]} : {prediction_Pokemon_1[0]:.4f}")
    print(f"{pokemon2[1]} : {prediction_Pokemon_2[0]:.4f}")
    if prediction_Pokemon_1[0] > prediction_Pokemon_2[0]:
        print(pokemon1[1].upper(), "EST LE VAINQUEUR !")
    elif prediction_Pokemon_2[0] > prediction_Pokemon_1[0]:
        print(pokemon2[1].upper(), "EST LE VAINQUEUR !")
    else:
        print("ÉGALITÉ !")
```

## 3. Combat d’exemple du tutoriel : 368 contre 598

```python
with open("datas/pokedex.csv", newline="", encoding="utf-8-sig") as csvfile:
    pokedex = list(csv.reader(csvfile))
    next(iter(pokedex))  # Le fichier commence par les en-têtes
    prediction(368, 598, pokedex[1:])
```
